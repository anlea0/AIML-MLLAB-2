```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import mean_squared_error, accuracy_score, r2_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, classification_report
```

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()
X = data.data
y = data.target
print("Classes:", data.target_names)

df = pd.DataFrame(X, columns=data.feature_names)
df['target'] = y

df.head()
df.tail()
df.columns
df.info()
df.describe()
```

```python
#Split Data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

#Linear Regression
lr = LogisticRegression(max_iter=5000)
lr.fit(X_train, y_train)
pred_lr = lr.predict(X_test)
print("Logistic TestAccuracy: ", accuracy_score(y_test, pred_lr))
pred_lr_train = lr.predict(X_train)
print("Logistic TestAccuracy: ", accuracy_score(y_train, pred_lr_train))
```

```python
#Decision Tree Classifier
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
pred_dt = dt.predict(X_test)
print("DT TestAccuracy: ", accuracy_score(y_test, pred_dt))
pred_dt_train = dt.predict(X_train)
print("DT TestAccuracy: ", accuracy_score(y_train, pred_dt_train))
```

```python
#KNeighbors Classifier
kn = KNeighborsClassifier()
kn.fit(X_train, y_train)
pred_kn = kn.predict(X_test)
print("KNN TestAccuracy: ", accuracy_score(y_test, pred_kn))
pred_kn_train = kn.predict(X_train)
print("KNN TestAccuracy: ", accuracy_score(y_train, pred_kn_train))
```

```python
#Naive-Bayes
nb = GaussianNB()
nb.fit(X_train, y_train)
pred_nb = nb.predict(X_test)
print("Naive-Bayes TestAccuracy: ", accuracy_score(y_test, pred_nb))
pred_nb_train = nb.predict(X_train)
print("Naive-Bayes TestAccuracy: ", accuracy_score(y_train, pred_nb_train))
```

```python
#Neural Network
nn = MLPClassifier(hidden_layer_sizes=(100,),
                   max_iter=1000,
                   random_state=42)
nn.fit(X_train, y_train)
pred_nn = nn.predict(X_test)
print("Neural Network TestAccuracy: ", accuracy_score(y_test, pred_nn))
pred_nn_train = nn.predict(X_train)
print("Neural Network TestAccuracy: ", accuracy_score(y_train, pred_nn_train))
```

```python
from sklearn.metrics import confusion_matrix
import pandas as pd

# 1. Define models list
classifiers = [
    ("Logistic Regression", lr),
    ("Decision Tree", dt),
    ("KNN", kn),
    ("Naive Bayes", nb),
    ("Neural Network", nn)
]

results_C = []

# 2. Iterate through models
for name, model in classifiers:
    y_pred = model.predict(X_test)
    cm = confusion_matrix(y_test, y_pred)
    
    # confusion_matrix structure for classes [0, 1]:
    # cm[0, 0]: Actual 0 (Malignant), Pred 0 (Malignant) -> Correct Disease Det.
    # cm[0, 1]: Actual 0 (Malignant), Pred 1 (Benign)    -> Type-II (False Negative)
    # cm[1, 0]: Actual 1 (Benign),    Pred 0 (Malignant) -> Type-I (False Positive)
    # cm[1, 1]: Actual 1 (Benign),    Pred 1 (Benign)    -> Correct Healthy
    
    # Extract counts
    true_malignant = cm[0, 0]
    false_negative = cm[0, 1] # Missed the disease
    false_positive = cm[1, 0] # False alarm
    true_benign    = cm[1, 1]
    
    # Calculate Rates
    # Type-I Rate = FP / (FP + TN) -> Proportion of Healthy predicted as Diseased
    type_I_rate = false_positive / (false_positive + true_benign) if (false_positive + true_benign) > 0 else 0
    
    # Type-II Rate = FN / (FN + TP) -> Proportion of Diseased predicted as Healthy
    type_II_rate = false_negative / (false_negative + true_malignant) if (false_negative + true_malignant) > 0 else 0
    
    # Observation Logic
    # In medical diagnosis, Type-II (Missing a disease) is more dangerous.
    if type_II_rate == 0:
        observation = "Ideal (No Missed Cases)"
    elif type_II_rate <= 0.05:
        observation = "Safe (Low Risk)"
    elif type_II_rate <= 0.10:
        observation = "Acceptable Risk"
    else:
        observation = "High Risk (Missed Diagnosis)"
        
    results_C.append({
        "Model": name,
        "Type-I Rate": round(type_I_rate, 4),
        "Type-II Rate": round(type_II_rate, 4),
        "Observation": observation
    })

# 3. Create DataFrame
table_c = pd.DataFrame(results_C)

print("Table C: Error analysis with Type-I and Type-II rates")
display(table_c)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score
import pandas as pd

# 1. Define models list (ensure these are trained as in previous steps)
classifiers = [
    ("Logistic Regression", lr),
    ("Decision Tree", dt),
    ("KNN", kn),
    ("Naive Bayes", nb),
    ("Neural Network", nn)
]

results_B = []

# 2. Iterate through models to calculate metrics
for name, model in classifiers:
    # Get predictions
    y_pred = model.predict(X_test)
    
    # Get probabilities for ROC-AUC (probability of positive class)
    # Most sklearn classifiers support predict_proba
    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1] 
    else:
        # Fallback if predict_proba is not available (unlikely for these standard models)
        y_prob = y_pred 
        
    # Calculate Metrics
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_prob)
    
    # 3. Apply Interpretation Logic based on Section 2.5 (AUC ranges)
    if auc < 0.6:
        interpretation = "Poor"
    elif 0.6 <= auc < 0.7:
        interpretation = "Acceptable"
    elif 0.7 <= auc < 0.8:
        interpretation = "Good"
    elif 0.8 <= auc < 0.9:
        interpretation = "Very Good"
    else: # auc >= 0.9
        interpretation = "Perfect/Excellent"
        
    # Append results
    results_B.append({
        "Model": name,
        "Precision": round(precision, 4),
        "Recall": round(recall, 4),
        "F1-score": round(f1, 4),
        "AUC/ROC": round(auc, 4),
        "Interpretation": interpretation
    })

# 4. Create DataFrame and Display
table_b = pd.DataFrame(results_B)

print("Table B: Comparison of Precision, Recall, F1-score and ROC-AUC")
display(table_b)
```
